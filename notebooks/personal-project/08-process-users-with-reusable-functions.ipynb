## Process all followed accounts (catch up)

### Get all followed account

```python
from typing import List
from utils.schemas import UserProfile

def get_all_followed_accounts(bluesky_client) -> List[UserProfile]:
    """
    Fetches all accounts followed by the authenticated user via the bluesky_client,
    using the app.bsky.graph.getFollows API conventions.

    Args:
        bluesky_client: An authenticated Bluesky client instance.

    Returns:
        List[UserProfile]: A list of UserProfile objects representing followed accounts.
    """
    # Try to determine the DID or handle of the current user gracefully
    user_identifier = (
        getattr(bluesky_client, "did", None)
        or getattr(bluesky_client, "handle", None)
    )
    
    # Fallback: check if the client can provide session info or user reference via another attribute or method
    if not user_identifier:
        # Try common Bluesky libraries: .me, .get_profile(), or .session
        did = None
        # Try .me
        if hasattr(bluesky_client, "me"):
            me = getattr(bluesky_client, "me")
            # me could be a profile dict, object or have .did
            if isinstance(me, dict) and "did" in me:
                did = me["did"]
            elif hasattr(me, "did"):
                did = me.did
            elif hasattr(me, "id"):
                did = me.id
        # Try .session
        if not did and hasattr(bluesky_client, "session"):
            session = getattr(bluesky_client, "session")
            if isinstance(session, dict) and "did" in session:
                did = session["did"]
            elif hasattr(session, "did"):
                did = session.did
        # Try .get_profile() with no parameter
        if not did and hasattr(bluesky_client, "get_profile"):
            try:
                profile = bluesky_client.get_profile()
                if isinstance(profile, dict) and "did" in profile:
                    did = profile["did"]
                elif hasattr(profile, "did"):
                    did = profile.did
            except Exception:
                pass
        # Try .user if exists (sometimes libraries set it like this)
        if not did and hasattr(bluesky_client, "user"):
            user = getattr(bluesky_client, "user")
            if isinstance(user, dict) and "did" in user:
                did = user["did"]
            elif hasattr(user, "did"):
                did = user.did
        if not did:
            raise ValueError(
                "Could not determine the current user did/handle from bluesky_client. "
                "Please check your Bluesky library for how to obtain the session DID or handle. "
                "Try passing the authenticated user's handle manually, or refer to your library's documentation."
            )
    else:
        # If handle is provided, try to resolve it to a DID
        if isinstance(user_identifier, str) and (user_identifier.startswith('@') or '.' in user_identifier):
            profile = bluesky_client.get_profile(user_identifier)
            did = getattr(profile, "did", user_identifier)
        else:
            did = user_identifier

    followed_accounts = []
    cursor = None

    while True:
        response = bluesky_client.get_follows(actor=did, limit=100, cursor=cursor)

        # Defensive handling: Refer to attributes, not dict methods (fixes AttributeError)
        # The response is likely a pydantic or object, not a dict.

        if hasattr(response, "follows"):
            follows = getattr(response, "follows") or []
        else:
            follows = []

        for user in follows:
            # user may be a dict or an object
            # Try attribute access, fall back to dict access for each field
            handle = getattr(user, "handle", None)
            if handle is None and isinstance(user, dict):
                handle = user.get("handle", "")
            handle = handle or ""
            display_name = getattr(user, "display_name", None)
            if display_name is None and hasattr(user, "displayName"):
                display_name = getattr(user, "displayName")
            if display_name is None and isinstance(user, dict):
                display_name = user.get("displayName", None)
            display_name = display_name or handle
            description = getattr(user, "description", None)
            if description is None and isinstance(user, dict):
                description = user.get("description", "")
            description = description or ""
            account_did = getattr(user, "did", None)
            if account_did is None and isinstance(user, dict):
                account_did = user.get("did", "")
            account_did = account_did or ""
            followed_accounts.append(UserProfile(
                handle=handle,
                display_name=display_name,
                description=description,
                did=account_did,
            ))

        # For cursor, do the same: attribute access, fallback to dict
        cursor_value = getattr(response, "cursor", None)
        if cursor_value is None and isinstance(response, dict):
            cursor_value = response.get("cursor")
        cursor = cursor_value
        if not cursor:
            break

    return followed_accounts
```

```python
followed_accounts = get_all_followed_accounts(bluesky_client)
print(f"Found {len(followed_accounts)} followed accounts")
# pretty print first 10 followed accounts
for account in followed_accounts[:10]:
    print(f"Handle: {account.handle}")
    print(f"Display Name: {account.display_name}")
    print(f"Description: {account.description}")
```

```python
process_users(followed_accounts,my_profile,openai_client,bluesky_client,qdrant_client)
print(f"Processed {len(followed_accounts)} users")
```
