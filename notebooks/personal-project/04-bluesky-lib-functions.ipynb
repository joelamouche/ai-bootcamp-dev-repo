```python
from utils.schemas import UserProfile
from typing import List
def get_custom_feed_profiles(bluesky_client,actor,feed_uri , limit=100)->List[UserProfile]: 
    profile = bluesky_client.app.bsky.actor.get_profile({"actor": actor})
    did = profile.did  # e.g. "did:plc:xyz..."

    # 3. Build the feed AT-URI from the URL
    # https://bsky.app/profile/skyfeed.xyz/feed/h-opensource
    feed_uri = f"at://{did}/app.bsky.feed.generator/{feed_uri}"

    # 4. Fetch the feed items
    resp = bluesky_client.app.bsky.feed.get_feed({"feed": feed_uri, "limit": limit})

    # get the profiles of the authors of the posts and
    # use client.get_profile to get the profile of the author   
    profiles = [bluesky_client.get_profile(item.post.author.handle) for item in resp.feed]
    # map the profiles to UserProfile
    profiles = [UserProfile(
        # if no field, set it to an empty string
        handle=profile.handle if profile.handle else "", 
        display_name=profile.display_name if profile.display_name else "", 
        description=profile.description if profile.description else "", 
        did=profile.did if profile.did else "") for profile in profiles]
    return profiles
```

```python
from utils.profile_sources import search_profiles_by_keyword
from utils.profile_filters import analyze_users_with_ai
from utils.profile_sources import get_my_profile
from utils.follow_profiles import follow_top_users
```

```python
# a function to retrieve all the posts in the feed of the user, in the last 24 hours
# the feed is the latets posts from the users the user follows
def get_my_feed(bluesky_client):
    # get the user profile
    #user_profile = get_my_profile(bluesky_client)
    # get the user feed using bluesky client
    feed = bluesky_client.get_timeline("following", None, 10)
    return feed

feed = get_my_feed(bluesky_client)
# Display first 3 posts in the feed (post text, author, and created_at)
# The 'feed' object is a Response, so we need to get its actual list of feed items.
feed_items = getattr(feed, "feed", None) or getattr(feed, "posts", None) or []
for i in range(min(10, len(feed_items))):
    post_item = feed_items[i]
    # Always handle post_item as an object, safely access attributes
    post = getattr(post_item, "post", None)
    if post is None:
        post = post_item  # Fallback: treat post_item itself as post if 'post' attribute missing

    # Safely extract author information
    author_obj = getattr(post, "author", None)
    if author_obj is not None:
        author = getattr(author_obj, "display_name", None) or getattr(author_obj, "handle", "")
    else:
        author = ""

    # Safely extract text and created_at from record (if exists)
    record = getattr(post, "record", None)
    if record is not None:
        text = getattr(record, "text", "")
        created_at = getattr(record, "created_at", "")
    else:
        text = ""
        created_at = ""

    print(f"---\nAuthor: {author}\nCreated at: {created_at}\n{text}\n")
```
