````python
# Import additional dependencies for AI analysis
from openai import OpenAI
import instructor
from pydantic import BaseModel, Field
from typing import List
import time

# Initialize OpenAI client with instructor for structured outputs
openai_client = instructor.from_openai(OpenAI())
````

````python
# Define structured output model for follower ranking
class FollowerRanking(BaseModel):
    """Ranked list of followers with relevance scores"""
    ranked_followers: List[dict] = Field(
        description="List of followers ranked by relevance, each with handle, display_name, description, and relevance_score"
    )

class RankedFollower(BaseModel):
    """Individual follower with relevance information"""
    handle: str = Field(description="Bluesky handle of the follower")
    display_name: str = Field(description="Display name of the follower")
    description: str = Field(description="Profile description of the follower")
    relevance_score: float = Field(description="Relevance score from 0.0 to 1.0, where 1.0 is most relevant")
    reasoning: str = Field(description="Brief explanation of why this follower is relevant")
````

````python
# Step 7 Functions: Reusable functions for follower analysis and following

from datetime import datetime, timedelta
import time

def is_user_active_in_last_month(client, user_did, days=30, debug=False):
    """
    Check if a user was active (posted) in the last N days.
    
    Args:
        client: Authenticated Bluesky client
        user_did: User's DID (Decentralized Identifier)
        days: Number of days to look back (default: 30)
        debug: If True, print debug information (default: False)
    
    Returns:
        bool: True if user posted in the last N days, False otherwise
    """
    try:
        # Calculate the cutoff date (using timezone-aware datetime)
        from datetime import timezone
        cutoff_date = datetime.now(timezone.utc) - timedelta(days=days)
        
        if debug:
            print(f"   Checking activity for {user_did}")
            print(f"   Cutoff date: {cutoff_date}")
        
        # Get user's recent posts (author feed)
        # Use the correct API method: client.app.bsky.feed.get_author_feed
        feed_response = client.app.bsky.feed.get_author_feed({'actor': user_did, 'limit': 10})
        
        if not hasattr(feed_response, 'feed') or not feed_response.feed:
            if debug:
                print(f"   No feed found or empty feed")
            return False
        
        if debug:
            print(f"   Found {len(feed_response.feed)} posts in feed")
        
        # Check if any post is within the last N days
        for i, feed_item in enumerate(feed_response.feed[:10]):  # Check first 10 posts
            try:
                # The feed structure in atproto: feed_item.post.record.createdAt
                created_at_str = None
                
                # Access the post object
                post_obj = feed_item.post if hasattr(feed_item, 'post') else None
                
                if post_obj is None:
                    if debug:
                        print(f"   Post {i}: No post object found")
                    continue
                
                # First, try to get indexed_at from post (snake_case, not camelCase)
                created_at_str = None
                if hasattr(post_obj, 'indexed_at'):
                    created_at_str = post_obj.indexed_at
                    if debug:
                        print(f"   Post {i}: Using indexed_at from post: {created_at_str}")
                elif hasattr(post_obj, 'indexedAt'):  # Fallback to camelCase just in case
                    created_at_str = post_obj.indexedAt
                    if debug:
                        print(f"   Post {i}: Using indexedAt from post: {created_at_str}")
                
                # If not found, try to get record from post - this is the key part
                # The record contains the actual post data including created_at (snake_case)
                if not created_at_str:
                    if hasattr(post_obj, 'record'):
                        record = post_obj.record
                    elif hasattr(post_obj, 'value'):
                        record = post_obj.value
                    else:
                        record = None
                    
                    if record is None:
                        if debug:
                            print(f"   Post {i}: No record found in post object")
                        continue
                    
                    # Get created_at from record (snake_case, not camelCase)
                    if isinstance(record, dict):
                        created_at_str = record.get('created_at', '') or record.get('createdAt', '')
                    else:
                        # Try attribute access with snake_case first
                        created_at_str = getattr(record, 'created_at', None)
                        if not created_at_str:
                            created_at_str = getattr(record, 'createdAt', None)  # Fallback to camelCase
                        
                        # If still None, try to convert record to dict
                        if created_at_str is None:
                            try:
                                if hasattr(record, 'model_dump'):
                                    record_dict = record.model_dump()
                                elif hasattr(record, 'dict'):
                                    record_dict = record.dict()
                                elif hasattr(record, '__dict__'):
                                    record_dict = record.__dict__
                                else:
                                    record_dict = {}
                                created_at_str = record_dict.get('created_at', '') or record_dict.get('createdAt', '')
                            except Exception as e:
                                if debug:
                                    print(f"   Post {i}: Error converting record to dict: {e}")
                                pass
                
                if not created_at_str:
                    if debug:
                        print(f"   Post {i}: Could not find created_at in record or indexed_at in post")
                    continue
                
                # Parse the timestamp (format: 2024-01-01T12:00:00.000Z)
                # Handle both with and without microseconds and timezone
                created_at_str_clean = str(created_at_str).replace('Z', '+00:00')
                if '.' not in created_at_str_clean.split('+')[0].split('-')[-1]:
                    # No microseconds, add them before the timezone
                    if '+00:00' in created_at_str_clean:
                        created_at_str_clean = created_at_str_clean.replace('+00:00', '.000+00:00')
                    elif 'Z' in str(created_at_str):
                        created_at_str_clean = str(created_at_str).replace('Z', '.000+00:00')
                
                created_at = datetime.fromisoformat(created_at_str_clean)
                # Ensure both datetimes are timezone-aware for comparison
                if created_at.tzinfo is None:
                    created_at = created_at.replace(tzinfo=timezone.utc)
                
                if debug:
                    print(f"   Post {i}: Created at {created_at}, Cutoff: {cutoff_date}, Active: {created_at >= cutoff_date}")
                
                if created_at >= cutoff_date:
                    if debug:
                        print(f"   ✅ User is active!")
                    return True
                    
            except Exception as e:
                if debug:
                    print(f"   Post {i}: Error processing - {e}")
                    import traceback
                    traceback.print_exc()
                continue
        
        if debug:
            print(f"   ❌ No recent posts found")
        return False
        
    except Exception as e:
        if debug:
            print(f"   ❌ Exception checking activity: {e}")
            import traceback
            traceback.print_exc()
        # If we can't check activity, assume inactive to be safe
        return False


def fetch_followers_from_user(client, username, max_followers=100, filter_active=True, active_days=30):
    """
    Fetch followers from a specific Bluesky user, optionally filtering by activity.
    
    Args:
        client: Authenticated Bluesky client
        username: Target username (with or without @)
        max_followers: Maximum number of followers to fetch (default: 100)
        filter_active: If True, only return followers active in the last month (default: True)
        active_days: Number of days to look back for activity (default: 30)
    
    Returns:
        tuple: (list of followers, handle_to_did mapping dict, target_profile)
    """
    # Clean username
    clean_username = username.lstrip('@').strip()
    
    print(f"📊 Fetching followers of @{clean_username}...")
    
    try:
        # Get the target profile
        target_profile = client.get_profile(clean_username)
        print(f"✅ Found profile: {target_profile.display_name or target_profile.handle}")
        print(f"   Followers: {target_profile.followers_count}")
        
        # Fetch followers (with pagination if needed)
        all_followers = []
        handle_to_did = {}
        cursor = None
        
        while len(all_followers) < max_followers:
            if cursor:
                followers_response = client.get_followers(target_profile.did, limit=100, cursor=cursor)
            else:
                followers_response = client.get_followers(target_profile.did, limit=100)
            
            all_followers.extend(followers_response.followers)
            
            # Build handle to DID mapping
            for follower in followers_response.followers:
                handle_to_did[follower.handle] = follower.did
            
            if not hasattr(followers_response, 'cursor') or not followers_response.cursor:
                break
            cursor = followers_response.cursor
            
            if len(all_followers) >= max_followers:
                break
        
        # Limit to the requested number (before filtering)
        all_followers = all_followers[:max_followers]
        
        print(f"✅ Fetched {len(all_followers)} followers")
        
        # Filter by activity if requested
        if filter_active:
            print(f"\n🔍 Filtering followers active in the last {active_days} days...")
            active_followers = []
            active_handle_to_did = {}
            checked_count = 0
            
            for follower in all_followers:
                checked_count += 1
                if checked_count % 10 == 0:
                    print(f"   Checking activity: {checked_count}/{len(all_followers)}...")
                
                if is_user_active_in_last_month(client, follower.did, days=active_days):
                    active_followers.append(follower)
                    active_handle_to_did[follower.handle] = follower.did
                
                # Small delay to avoid rate limiting
                time.sleep(0.1)
            
            print(f"✅ Found {len(active_followers)} active followers out of {len(all_followers)} checked")
            return active_followers, active_handle_to_did, target_profile
        else:
            return all_followers, handle_to_did, target_profile
        
    except Exception as e:
        print(f"❌ Error fetching followers: {e}")
        import traceback
        traceback.print_exc()
        return [], {}, None


def get_my_profile(client):
    """
    Get your own profile information.
    
    Args:
        client: Authenticated Bluesky client
    
    Returns:
        Profile object with your profile information
    """
    my_profile = client.get_profile(client.me.handle)
    return my_profile


def analyze_followers_with_ai(followers, my_profile, target_username, openai_client, num_results=10):
    """
    Use AI to analyze and rank followers by relevance to your profile.
    
    Args:
        followers: List of follower objects
        my_profile: Your profile object
        target_username: Username of the target account whose followers we're analyzing
        openai_client: OpenAI client with instructor
        num_results: Number of top results to return (default: 10)
    
    Returns:
        List of RankedFollower objects sorted by relevance
    """
    if not followers:
        print("⚠️ No followers to analyze")
        return []
    
    print(f"\n🤖 Analyzing {len(followers)} followers with AI...")
    
    # Prepare follower data for AI analysis
    followers_data = []
    for follower in followers:
        followers_data.append({
            "handle": follower.handle,
            "display_name": follower.display_name or follower.handle,
            "description": follower.description or ""
        })
    
    # Get your profile info
    my_description = my_profile.description or ""
    my_display_name = my_profile.display_name or my_profile.handle
    
    # Create prompt for AI analysis
    analysis_prompt = f"""You are analyzing Bluesky followers to find the most relevant accounts for someone to follow.

My Profile:
- Display Name: {my_display_name}
- Handle: {my_profile.handle}
- Description: {my_description}

I want to find followers from @{target_username}'s account who would be most relevant for me to follow based on:
1. Shared interests (based on my description: {my_description[:200]})
2. Professional alignment
3. Content relevance
4. Community overlap

Here are the followers to analyze:
{chr(10).join([f"- @{f['handle']} ({f['display_name']}): {f['description'][:150]}" for f in followers_data[:50]])}

Please rank these followers by relevance to my profile. Return the top {num_results} most relevant followers with:
- handle: The Bluesky handle
- display_name: Their display name
- description: Their profile description
- relevance_score: A score from 0.0 to 1.0 (1.0 = most relevant)
- reasoning: A brief explanation of why they're relevant

Focus on accounts that share interests, professional connections, or would provide valuable content for someone with my profile."""
    
    try:
        # Use instructor to get structured output
        response = openai_client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "You are an expert at analyzing social media profiles and finding relevant connections."},
                {"role": "user", "content": analysis_prompt}
            ],
            response_model=List[RankedFollower],
            temperature=0.3
        )
        
        ranked_followers = response
        print(f"✅ AI analysis complete! Found {len(ranked_followers)} relevant followers")
        
        return ranked_followers
        
    except Exception as e:
        print(f"❌ Error during AI analysis: {e}")
        import traceback
        traceback.print_exc()
        return []


def get_top_followers_from_user(username, number_of_followers, client=None, openai_client=None, max_followers_to_analyze=100, filter_active=True, active_days=30):
    """
    Main function: Get top relevant followers from a specific user.
    
    This function:
    1. Fetches followers from the target user (optionally filtering by activity)
    2. Gets your profile for context
    3. Uses AI to analyze and rank followers
    4. Returns the top N most relevant followers
    
    Args:
        username: Target username (with or without @)
        number_of_followers: Number of top followers to return
        client: Bluesky client (uses global 'client' if not provided)
        openai_client: OpenAI client with instructor (uses global 'openai_client' if not provided)
        max_followers_to_analyze: Maximum followers to fetch and analyze (default: 100)
        filter_active: If True, only analyze followers active in the last month (default: True)
        active_days: Number of days to look back for activity (default: 30)
    
    Returns:
        tuple: (list of RankedFollower objects, handle_to_did mapping dict)
    """
    # Use global clients if not provided
    if client is None:
        client = globals().get('client')
        if client is None:
            raise ValueError("Bluesky client not provided and not found in globals. Please authenticate first.")
    
    if openai_client is None:
        openai_client = globals().get('openai_client')
        if openai_client is None:
            raise ValueError("OpenAI client not provided and not found in globals. Please initialize first.")
    
    # Step 1: Fetch followers (with activity filtering)
    all_followers, handle_to_did, target_profile = fetch_followers_from_user(
        client, username, max_followers=max_followers_to_analyze, 
        filter_active=filter_active, active_days=active_days
    )
    
    if not all_followers:
        print("⚠️ No followers found")
        return [], {}
    
    # Step 2: Get your profile
    my_profile = get_my_profile(client)
    print(f"\n📝 Your profile context:")
    print(f"   Name: {my_profile.display_name or my_profile.handle}")
    print(f"   Handle: {my_profile.handle}")
    print(f"   Description: {(my_profile.description or '')[:200]}...")
    
    # Step 3: Analyze with AI
    ranked_followers = analyze_followers_with_ai(
        all_followers, 
        my_profile, 
        username.lstrip('@').strip(),
        openai_client,
        num_results=number_of_followers
    )
    
    # Display results
    if ranked_followers:
        print(f"\n🏆 Top {min(number_of_followers, len(ranked_followers))} Most Relevant Followers:")
        for i, follower in enumerate(ranked_followers[:number_of_followers], 1):
            print(f"\n{i}. @{follower.handle} ({follower.display_name})")
            print(f"   Relevance Score: {follower.relevance_score:.2f}")
            print(f"   Reasoning: {follower.reasoning}")
            print(f"   Description: {follower.description[:100]}...")
    
    return ranked_followers[:number_of_followers], handle_to_did


def follow_top_followers(ranked_followers, client=None, handle_to_did=None, top_n=10):
    """
    Follow the top N ranked followers.
    
    Args:
        ranked_followers: List of RankedFollower objects
        client: Bluesky client (uses global 'client' if not provided)
        handle_to_did: Mapping from handle to DID (optional, will fetch if not provided)
        top_n: Number of top followers to follow (default: 10)
    
    Returns:
        dict: Summary with counts of followed, already_following, errors, etc.
    """
    if not ranked_followers:
        print("⚠️ No ranked followers to follow")
        return {}
    
    # Use global client if not provided
    if client is None:
        client = globals().get('client')
        if client is None:
            raise ValueError("Bluesky client not provided and not found in globals.")
    
    top_n_followers = ranked_followers[:top_n]
    print(f"\n👥 Following top {len(top_n_followers)} most relevant accounts...")
    
    followed_count = 0
    already_following_count = 0
    error_count = 0
    not_found_count = 0
    
    for i, follower in enumerate(top_n_followers, 1):
        try:
            # if follower relevance score is less than 0.5, skip
            if follower.relevance_score < 0.55:
                print(f"ℹ️  {i}. Skipping @{follower.handle} ({follower.display_name}) due to low relevance score: {follower.relevance_score}")
                continue

            # Clean handle: remove '@' if present and normalize
            clean_handle = follower.handle.lstrip('@').strip()
            
            # Get the DID
            follower_did = None
            if handle_to_did:
                follower_did = handle_to_did.get(clean_handle)
            
            if not follower_did:
                # Fetch profile to get DID
                try:
                    profile = client.get_profile(clean_handle)
                    follower_did = profile.did
                except Exception as e:
                    not_found_count += 1
                    print(f"⚠️  {i}. Could not find DID for @{clean_handle}: {e}")
                    continue
            
            # Follow using the DID (not the handle)
            client.follow(follower_did)
            followed_count += 1
            print(f"✅ {i}. Followed @{clean_handle} ({follower.display_name})")
            
            # Add a small delay to avoid rate limiting
            time.sleep(1)
            
        except Exception as e:
            error_count += 1
            error_msg = str(e)
            clean_handle = follower.handle.lstrip('@').strip()
            
            if "already" in error_msg.lower() or "duplicate" in error_msg.lower():
                already_following_count += 1
                print(f"ℹ️  {i}. Already following @{clean_handle}")
            else:
                print(f"❌ {i}. Error following @{clean_handle}: {error_msg}")
    
    summary = {
        'followed': followed_count,
        'already_following': already_following_count,
        'not_found': not_found_count,
        'errors': error_count,
        'total_processed': len(top_n_followers)
    }
    
    print(f"\n📊 Summary:")
    print(f"   ✅ Successfully followed: {followed_count}")
    print(f"   ℹ️  Already following: {already_following_count}")
    print(f"   ⚠️  Not found: {not_found_count}")
    print(f"   ❌ Errors: {error_count}")
    print(f"   📈 Total processed: {len(top_n_followers)}")
    
    return summary

print("✅ All Step 7 functions loaded!")
````

### Step 7: Expanding Follower Base with AI Analysis

This step demonstrates how to:
1. Fetch followers from a specific profile
2. Use AI (ChatGPT) to analyze which followers would be most relevant to follow based on your profile
3. Automatically follow the top 10 most relevant accounts

**Use case:** Find and follow relevant accounts from someone who has a similar audience or interests.


````python
# Import additional dependencies for AI analysis
from openai import OpenAI
import instructor
from pydantic import BaseModel, Field
from typing import List
import time

# Initialize OpenAI client with instructor for structured outputs
openai_client = instructor.from_openai(OpenAI())
````

````python
# Define structured output model for follower ranking
class FollowerRanking(BaseModel):
    """Ranked list of followers with relevance scores"""
    ranked_followers: List[dict] = Field(
        description="List of followers ranked by relevance, each with handle, display_name, description, and relevance_score"
    )

class RankedFollower(BaseModel):
    """Individual follower with relevance information"""
    handle: str = Field(description="Bluesky handle of the follower")
    display_name: str = Field(description="Display name of the follower")
    description: str = Field(description="Profile description of the follower")
    relevance_score: float = Field(description="Relevance score from 0.0 to 1.0, where 1.0 is most relevant")
    reasoning: str = Field(description="Brief explanation of why this follower is relevant")
````

````python
# Step 7.1: Fetch followers from a target profile
target_username = 'simondlr.com'  # Change this to the profile whose followers you want to analyze

print(f"📊 Fetching followers of @{target_username}...")

try:
    # Get the target profile
    target_profile = client.get_profile(target_username)
    print(f"✅ Found profile: {target_profile.display_name or target_profile.handle}")
    print(f"   Followers: {target_profile.followers_count}")
    
    # Fetch followers (with pagination if needed)
    # Note: get_followers returns a limited number per call, you may need to paginate
    all_followers = []
    cursor = None
    max_followers_to_fetch = 100  # Limit to avoid rate limits and keep analysis manageable
    
    while len(all_followers) < max_followers_to_fetch:
        if cursor:
            followers_response = client.get_followers(target_profile.did, limit=100, cursor=cursor)
        else:
            followers_response = client.get_followers(target_profile.did, limit=100)
        
        all_followers.extend(followers_response.followers)
        
        if not hasattr(followers_response, 'cursor') or not followers_response.cursor:
            break
        cursor = followers_response.cursor
        
        if len(all_followers) >= max_followers_to_fetch:
            break
    
    # Limit to the requested number
    all_followers = all_followers[:max_followers_to_fetch]
    
    print(f"✅ Fetched {len(all_followers)} followers")
    
except Exception as e:
    print(f"❌ Error fetching followers: {e}")
    import traceback
    traceback.print_exc()
    all_followers = []
````

````python
# Step 7.2: Get your own profile description for context
my_profile = client.get_profile(client.me.handle)
my_description = my_profile.description or ""
my_display_name = my_profile.display_name or my_profile.handle

print(f"📝 Your profile context:")
print(f"   Name: {my_display_name}")
print(f"   Handle: {my_profile.handle}")
print(f"   Description: {my_description[:200]}...")
````

````python
# Step 7.3: Use AI to analyze and rank followers
if all_followers:
    print(f"\n🤖 Analyzing {len(all_followers)} followers with AI...")
    
    # Create a mapping from handle to DID (needed for following)
    handle_to_did = {}
    for follower in all_followers:
        handle_to_did[follower.handle] = follower.did
    
    # Prepare follower data for AI analysis
    followers_data = []
    for follower in all_followers:
        followers_data.append({
            "handle": follower.handle,
            "display_name": follower.display_name or follower.handle,
            "description": follower.description or ""
        })
    
    # Create prompt for AI analysis
    analysis_prompt = f"""You are analyzing Bluesky followers to find the most relevant accounts for someone to follow.

My Profile:
- Display Name: {my_display_name}
- Handle: {my_profile.handle}
- Description: {my_description}

I want to find followers from @{target_username}'s account who would be most relevant for me to follow based on:
1. Shared interests (based on my description: {my_description[:200]})
2. Professional alignment
3. Content relevance
4. Community overlap

Here are the followers to analyze:
{chr(10).join([f"- @{f['handle']} ({f['display_name']}): {f['description'][:150]}" for f in followers_data[:50]])}

Please rank these followers by relevance to my profile. Return the top 10 most relevant followers with:
- handle: The Bluesky handle
- display_name: Their display name
- description: Their profile description
- relevance_score: A score from 0.0 to 1.0 (1.0 = most relevant)
- reasoning: A brief explanation of why they're relevant

Focus on accounts that share interests, professional connections, or would provide valuable content for someone with my profile."""
    
    try:
        # Use instructor to get structured output
        response = openai_client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "You are an expert at analyzing social media profiles and finding relevant connections."},
                {"role": "user", "content": analysis_prompt}
            ],
            response_model=List[RankedFollower],
            temperature=0.3
        )
        
        ranked_followers = response
        print(f"✅ AI analysis complete! Found {len(ranked_followers)} relevant followers")
        
        # Display top 10
        print(f"\n🏆 Top {min(10, len(ranked_followers))} Most Relevant Followers:")
        for i, follower in enumerate(ranked_followers[:10], 1):
            print(f"\n{i}. {follower.handle} ({follower.display_name})")
            print(f"   Relevance Score: {follower.relevance_score:.2f}")
            print(f"   Reasoning: {follower.reasoning}")
            print(f"   Description: {follower.description[:100]}...")
        
    except Exception as e:
        print(f"❌ Error during AI analysis: {e}")
        import traceback
        traceback.print_exc()
        ranked_followers = []
else:
    print("⚠️ No followers to analyze")
    ranked_followers = []
````

````python
# Step 7.4: Follow the top 10 most relevant accounts
if ranked_followers:
    top_10 = ranked_followers[:10]
    print(f"\n👥 Following top {len(top_10)} most relevant accounts...")
    
    followed_count = 0
    already_following_count = 0
    error_count = 0
    not_found_count = 0
    
    # Create or recreate handle_to_did mapping if not available
    # (in case this cell is run independently)
    if 'handle_to_did' not in globals() or not handle_to_did:
        handle_to_did = {}
        if 'all_followers' in globals() and all_followers:
            for follower in all_followers:
                handle_to_did[follower.handle] = follower.did
    
    for i, follower in enumerate(top_10, 1):
        try:
            # Clean handle: remove '@' if present and normalize
            clean_handle = follower.handle.lstrip('@').strip()
            
            # Get the DID - first try mapping, then fetch profile if needed
            follower_did = None
            if 'handle_to_did' in globals() and handle_to_did:
                follower_did = handle_to_did.get(clean_handle)
            
            if not follower_did:
                # Fetch profile to get DID
                try:
                    profile = client.get_profile(clean_handle)
                    follower_did = profile.did
                except Exception as e:
                    not_found_count += 1
                    print(f"⚠️  {i}. Could not find DID for @{clean_handle}: {e}")
                    continue
            
            # Follow using the DID (not the handle)
            client.follow(follower_did)
            followed_count += 1
            print(f"✅ {i}. Followed @{clean_handle} ({follower.display_name})")
            
            # Add a small delay to avoid rate limiting
            time.sleep(1)
            
        except Exception as e:
            error_count += 1
            error_msg = str(e)
            clean_handle = follower.handle.lstrip('@').strip()
            
            if "already" in error_msg.lower() or "duplicate" in error_msg.lower():
                already_following_count += 1
                print(f"ℹ️  {i}. Already following @{clean_handle}")
            else:
                print(f"❌ {i}. Error following @{clean_handle}: {error_msg}")
    
    print(f"\n📊 Summary:")
    print(f"   ✅ Successfully followed: {followed_count}")
    print(f"   ℹ️  Already following: {already_following_count}")
    print(f"   ⚠️  Not found: {not_found_count}")
    print(f"   ❌ Errors: {error_count}")
    print(f"   📈 Total processed: {len(top_10)}")
else:
    print("⚠️ No ranked followers to follow")
````
